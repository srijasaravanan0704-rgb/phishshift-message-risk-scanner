```python
import pypdf
import pandas as pd
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# 1. Extract text from the PDF file
pdf_path = '/content/ML-07_PhishShift_Message_Risk_Scanner_View_Only.pdf'
reader = pypdf.PdfReader(pdf_path)

pages_data = []
for i, page in enumerate(reader.pages):
    text = page.extract_text() or ""
    pages_data.append({
        'page_number': i + 1,
        'text': text.strip()
    })

df = pd.DataFrame(pages_data)
print("Extracted pages:", len(df))
print("Total characters extracted:", sum(df['text'].str.len()))
```

### Label Generation and Model Training
Since we are building a message risk scanner, we will segment the extracted text into smaller message-like chunks and simulate/assign risk labels (e.g., "Safe" vs "Phishing" / "High Risk") to train our Logistic Regression model. Then we will export the model components (`TfidfVectorizer` and the `LogisticRegression` model).

```python
# 2. Prepare text segments and assign risk labels. Fallback to synthetic dataset if PDF text is empty.
segments = []
for idx, row in df.iterrows():
    lines = row['text'].split('\n')
    for line in lines:
        cleaned_line = line.strip()
        if len(cleaned_line) > 10:
            segments.append(cleaned_line)

# Fallback to realistic phishing scanner dataset if extraction yields empty rows
if len(segments) < 10:
    print("Warning: Little or no text extracted from PDF. Using synthetic PhishShift dataset fallback for training...")
    synthetic_data = [
        ("URGENT: Your account has been suspended. Click here to verify your login credentials.", 1),
        ("Dear customer, please review your recent invoice attached to this email immediately.", 1),
        ("Congratulations! You have won a $1000 Amazon Gift Card. Claim your reward now!", 1),
        ("Warning: Security alert detected on your profile. Update password at secure-link.com.", 1),
        ("Verify your bank account details within 24 hours to avoid service interruption.", 1),
        ("Hey, are we still on for lunch tomorrow? Let me know.", 0),
        ("The weekly progress report has been uploaded to the shared folder.", 0),
        ("Please find the meeting minutes from our discussion yesterday attached.", 0),
        ("Happy birthday! Wishing you a fantastic year ahead!", 0),
        ("Your package has been shipped and is expected to arrive on Wednesday.", 0),
        ("Update: The system maintenance schedule has been updated to next Sunday.", 0),
        ("Suspicious activity detected on your account. Reset your secure password here.", 1),
        ("Reminder: Team sync meeting at 10:00 AM in Conference Room B.", 0),
        ("Final Warning: Your storage is full. Upgrade now to preserve your data.", 1),
        ("Can you send me the guidelines for the new project proposal? Thanks.", 0)
    ]
    df_messages = pd.DataFrame(synthetic_data, columns=['message', 'label'])
else:
    df_messages = pd.DataFrame(segments, columns=['message'])
    risk_keywords = ['risk', 'phish', 'security', 'alert', 'scanner', 'suspicious', 'urgent', 'password', 'login', 'verify', 'update']
    df_messages['label'] = df_messages['message'].apply(
        lambda x: 1 if any(keyword in x.lower() for keyword in risk_keywords) else 0
    )

print(f"Dataset size: {len(df_messages)} samples")
print(df_messages['label'].value_counts())
display(df_messages.head(10))
```

```python
# 3. Split data, train vectorizer and Logistic Regression model
X_train, X_test, y_train, y_test = train_test_split(
    df_messages['message'], df_messages['label'], test_size=0.2, random_state=42, stratify=df_messages['label']
)

vectorizer = TfidfVectorizer(max_features=1000, stop_words='english')
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

model = LogisticRegression()
model.fit(X_train_tfidf, y_train)

# Evaluate the model
y_pred = model.predict(X_test_tfidf)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
# 4. Export the model and vectorizer files
import joblib
import os

joblib.dump(model, 'logistic_regression_model.pkl')
joblib.dump(vectorizer, 'tfidf_vectorizer.pkl')

print("Successfully exported:")
print("- logistic_regression_model.pkl (Exists:", os.path.exists('logistic_regression_model.pkl'), ")")
print("- tfidf_vectorizer.pkl (Exists:", os.path.exists('tfidf_vectorizer.pkl'), ")")
```

```python
import joblib
import os
from flask import Flask, request, jsonify

app = Flask(__name__)

# Load the exported models
MODEL_PATH = 'logistic_regression_model.pkl'
VECTORIZER_PATH = 'tfidf_vectorizer.pkl'

if os.path.exists(MODEL_PATH) and os.path.exists(VECTORIZER_PATH):
    model = joblib.load(MODEL_PATH)
    vectorizer = joblib.load(VECTORIZER_PATH)
    print("Model and Vectorizer loaded successfully!")
else:
    print("Error: Model or Vectorizer files missing. Please run the training cells first.")

@app.route('/')
def home():
    return "Welcome to the PhishShift Message Risk Scanner API! Use the /predict endpoint to scan messages."

@app.route('/predict', methods=['POST'])
def predict():
    try:
        data = request.get_json(force=True)
        if not data or 'message' not in data:
            return jsonify({'error': 'Please provide a "message" field in the request body.'}), 400

        message = data['message']
        # Transform input using the vectorizer
        message_tfidf = vectorizer.transform([message])

        # Predict risk category
        prediction = int(model.predict(message_tfidf)[0])
        probability = model.predict_proba(message_tfidf)[0]
        risk_prob = float(probability[1])

        result = {
            'message': message,
            'is_high_risk': prediction,
            'risk_score': risk_prob,
            'status': 'Phishing / High Risk' if prediction == 1 else 'Safe'
        }
        return jsonify(result)
    except Exception as e:
        return jsonify({'error': str(e)}), 500

if __name__ == '__main__':
    # Note: To run Flask inside Google Colab, we run it on localhost port 5000.
    # You can interact with it locally using threading or expose it via ngrok/localtunnel.
    print("Starting Flask development server...")
    # app.run(host='0.0.0.0', port=5000)
```

### Testing your Flask API locally
To test this API directly inside the Colab environment, we can launch the server in a separate thread and send a POST request using python's `requests` library.

```python
import threading
import requests
import time

# Run the Flask app in a separate background thread
def run_app():
    app.run(host='127.0.0.1', port=5000, debug=False, use_reloader=False)

thread = threading.Thread(target=run_app)
thread.daemon = True
thread.start()

# Give the server 2 seconds to start
time.sleep(2)
```
